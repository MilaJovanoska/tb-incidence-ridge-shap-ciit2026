## Global Health Challenge, Biological Basis and Worldwide Impact

<p style="text-align: justify;">
<b>Tuberculosis (TB)</b> is a <b>chronic infectious disease</b> caused by the bacterium <i>Mycobacterium tuberculosis</i>, first identified in <b>1882</b> by <b>Dr. Robert Koch</b>, whose discovery marked a <b>groundbreaking advance</b> in medical microbiology. Although the <b>lungs</b> are the primary site of infection, TB can also affect other organs such as the <b>bones</b>, <b>brain</b>, and <b>kidneys</b>. This ability to spread beyond the respiratory system underscores the <b>clinical complexity</b> and <b>severity</b> of the disease, as it is not confined to a single organ.
</p>

<p style="text-align: justify;">
Despite being both <b>preventable</b> and <b>curable</b>, tuberculosis (TB) continues to represent a <b>major global health challenge</b>. Each year, millions of people become newly infected, and a significant number of deaths occur due to <b>delayed diagnosis</b> or <b>incomplete treatment</b>.
</p>

## Transmission Dynamics and Pathogenesis

<p style="text-align: justify;">
  The predominant mode of tuberculosis (TB) transmission is the <b>inhalation of airborne droplets</b>, 
  which are microscopic particles containing <i>Mycobacterium tuberculosis</i>, generated when an individual 
  with <b>active pulmonary tuberculosis (ATB)</b> coughs, sneezes, speaks, or even sings. These droplets 
  rapidly dry into <b>droplet nuclei</b> - tiny residual particles (1–5 &mu;m in diameter) - that can remain 
  suspended in the air for prolonged periods and reach the deepest parts of the lungs upon inhalation.
</p>

<ul style="text-align: justify; margin-left: 20px;">
  <li>
    A single cough may release <b>thousands of infectious particles</b>, and inhalation of only a few bacilli 
    (bacterial cells) can be sufficient to establish infection.
  </li>
  <li>
    TB is <b>not transmitted</b> through casual physical contact (such as handshakes, hugging, or sharing food 
    and utensils).
  </li>
  <li>
    <b>Prolonged and close exposure</b> in <b>confined or poorly ventilated environments</b> - such as households, 
    overcrowded healthcare facilities, or correctional institutions - is typically required for transmission to occur.
  </li>
</ul>

<p style="text-align: justify;">
  To contextualize the global burden of tuberculosis, it is essential to distinguish between its primary 
  clinical forms. A clear understanding of this distinction provides the foundation for 
  interpreting patterns of transmission, evaluating risk factors, and designing appropriate public-health 
  interventions.
</p>

<p style="text-align: justify;">
    There are two main forms of tuberculosis: 
</p>

<ul style="margin-left: 20px; text-align: justify;">
  <li>
    <b>Latent TB Infection (LTBI):</b>
    Individuals are infected with <i>Mycobacterium tuberculosis</i>, but the bacteria remain in a
    <b>dormant state</b>. At this stage, there are <b>no clinical symptoms</b>, and the infection
    <b>cannot be transmitted</b> to others.
  </li>

  <li style="margin-top: 6px;">
    <b>Active TB Disease (ATB):</b>
    The bacteria become <b>active and multiply</b>, producing <b>characteristic clinical symptoms</b>
    such as a persistent cough, fever, night sweats, and weight loss. In this form, the disease is
    <b>infectious</b>, and affected individuals can <b>transmit the infection</b> to others.
  </li>
</ul>


<p style="text-align: justify;">
  Understanding these differences is essential for <b>effective diagnosis</b>, <b>treatment strategies</b>, and 
  <b>public health planning</b>, as both <b>latent</b> and <b>active TB</b> contribute significantly to the 
  <b>global burden of the disease</b>.
</p>

## Treatment and Drug Resistance

<p style="text-align: justify;">
  Tuberculosis (TB) treatment relies on a <b>long, multidrug regimen</b> because 
  <i>Mycobacterium tuberculosis</i> grows slowly and can easily develop 
  <b>drug resistance</b> if therapy is interrupted. Drug-sensitive TB is typically 
  treated for <b>six months</b> using a combination of first-line antibiotics, 
  achieving cure rates of <b>up to 95%</b> with full adherence. Ensuring patients 
  complete treatment is critical, and strategies such as 
  <b>Directly Observed Therapy (DOT)</b> are widely used to prevent missed doses.
</p>

<p style="text-align: justify;">
  A major global challenge is the rise of <b>drug-resistant TB</b>, especially 
  <b>multidrug-resistant TB (MDR-TB)</b> and <b>extensively drug-resistant TB (XDR-TB)</b>, 
  which require longer, more toxic, and less effective treatments, often with cure rates 
  below <b>60%</b>. Recent advances - including new drugs such as <b>bedaquiline</b>, 
  <b>delamanid</b>, and <b>pretomanid</b>, and shorter <b>six-month all-oral regimens</b> 
  (e.g., BPaL/BPaLM) - have significantly improved outcomes for MDR-TB. Emerging 
  <b>four-month regimens</b> for drug-sensitive TB also show promise in improving 
  adherence and reducing treatment duration.
</p>

## Socio-Economic Impact

<p style="text-align: justify;">
Tuberculosis (TB) is not only a medical condition but also a major <b>socio-economic</b> burden, 
with consequences for households, labour markets, and national development.
</p>

<ul>
  <li><b>Catastrophic household costs:</b> Around half of TB-affected households incur
      “<i>catastrophic</i>” TB-related expenses (typically &gt;<b>20%</b> of annual income), especially in
      case of multidrug-resistant TB, pushing many families into or deeper into poverty.</li>
  <li style="margin-top: 6px;"><b>Loss of income and productivity:</b> TB mainly affects adults in their most productive
      years and often leads to several months of work lost, reduced earning capacity, and 
      long-term employment disruption.</li>
  <li style="margin-top: 6px;"><b>Macroeconomic losses:</b> Cumulatively, TB-related illness and premature death are 
      estimated to cause hundreds of billions of US dollars in lost economic output; between 
      <b>2015</b> and <b>2030</b> alone, projections suggest about <b>28 million</b> deaths and nearly 
      <b>1 trillion US$</b> in global economic losses if control efforts are not intensified.</li>
  <li style="margin-top: 6px;"><b>Reinforcing the poverty–TB cycle:</b> These costs, combined with stigma and social
      exclusion, create a self-perpetuating cycle in which poverty increases TB risk and TB
      in turn deepens poverty and inequality.</li>
</ul>

<p style="text-align: justify;">
Given this tight linkage between TB and structural conditions such as poverty, food
insecurity, and undernutrition, the next section shifts from conceptual discussion to

</p>

### Cleaning and Standardizing Country Names

<p style="text-align: justify;">
  Before analysis and dataset integration, we correct <b>inconsistent</b> and <b>incorrect</b> country labels in the FAOSTAT
  <b>Area</b> field. This step removes redundant territorial entries that can distort results and then standardizes remaining
  names to a single, consistent format for reliable grouping and merging.

```python
faostat_dataset["Area"] = faostat_dataset["Area"].astype(str).str.strip()

drop_areas = ["China", "China, Hong Kong SAR", "China, Macao SAR"]
faostat_dataset = faostat_dataset[~faostat_dataset["Area"].isin(drop_areas)].copy()

faostat_dataset["Area"] = faostat_dataset["Area"].replace({
    "China, mainland": "China",
    "China, Taiwan Province of": "Taiwan",
})
```

<p style="text-align: justify;">
  First, whitespace is stripped from the <b>Area</b> column to avoid hidden duplicates. Next, specific problematic records
  (e.g., separate SAR entries) are filtered out to prevent double counting. Finally, remaining variants are renamed into a
  consistent naming scheme (e.g., aligning “China, mainland” to “China” and “China, Taiwan Province of” to “Taiwan”) so that
  all observations map cleanly to the intended country categories.

    (<i>&micro;g/cap/d</i>).
  </li>
  <li>
    <b>Value</b> – Numerical observation, showing the amount of the selected indicator for a

  <li>

  </li>

    empty.
  </li>
</ol>

### Global Distribution of Dietary Energy Supply

<p style="text-align: justify;">
  The map <b>“Global Dietary Energy Supply (kcal per capita per day)”</b> shows each country’s
  <b>max daily dietary energy availability</b> (<i>kcal/capita/day</i>) aggregated across all
  FAOSTAT food groups. <b>Darker green</b> indicates <b>higher</b> energy availability, while
  <b>lighter green</b> indicates <b>lower</b> availability. Importantly, this is a measure of
  <b>food supply/availability</b>, not directly a measurment of an <b>actual calorie intake</b>, but it is
  widely used as a practical proxy for national nutrition conditions.
</p>

```python
def m49_to_iso3(m49):
    if pd.isna(m49):
        return None
    m49_str = f"{int(m49):03d}"
    c = pycountry.countries.get(numeric=m49_str)
    return getattr(c, "alpha_3", None)

faostat_dataset = faostat_dataset.copy()

codes = faostat_dataset["Area Code (M49)"].dropna().astype(int).unique()
m49_map = {code: m49_to_iso3(code) for code in codes}

faostat_dataset["iso3"] = (
    faostat_dataset["Area Code (M49)"].astype("Int64").map(m49_map)
)
```

```python
energy_supply = faostat_dataset[faostat_dataset["Indicator"] == "Energy supply"].copy()

energy_per_capita = (
    faostat_dataset.loc[faostat_dataset["Indicator"].eq("Energy supply")]
    .dropna(subset=["iso3"])
    .groupby("iso3", as_index=False)
    .agg(
        Area=("Area", "first"),
        energy_kcal_per_capita_day=("Value", "sum")
    )
)

green_scale = [
    (0.0, "#f7fcf5"),
    (0.2, "#e5f5e0"),
    (0.4, "#c7e9c0"),
    (0.6, "#a1d99b"),
    (0.8, "#74c476"),
    (1.0, "#006d2c"),
]

fig = px.choropleth(
    energy_per_capita,
    locations="iso3",
    locationmode="ISO-3",
    color="energy_kcal_per_capita_day",
    hover_name="Area",
    labels={"energy_kcal_per_capita_day": "kcal/capita/day"},
    title="Global Dietary Energy Supply (kcal per capita per day)",
    color_continuous_scale=green_scale,
    custom_data=["energy_kcal_per_capita_day"],
)

fig.update_traces(
    hovertemplate=(
        "<span style='font-size:16px;'><b>%{hovertext}</b></span><br>"
        "<span style='color:#888;'>kcal per capita per day</span><br>"
        "<span style='font-size:22px; color:#006d2c;'><b>%{customdata[0]:.0f}</b></span>"
        "<extra></extra>"
    ),
    marker_line_color="white",
    marker_line_width=0.4,
)

fig.update_layout(
    template="plotly_white",
    font=dict(family="Arial, sans-serif", size=13, color="#333333"),
    hoverlabel=dict(
        bgcolor="white",
        bordercolor="#cfcfcf",
        font_size=13,
        font_family="Arial, sans-serif",
        align="left",
    ),
    coloraxis_colorbar=dict(
        title="kcal/capita/day",
        ticks="outside",
        ticklen=6,
        outlinewidth=0,
    ),
    margin=dict(l=10, r=60, t=60, b=10),
    height=550,
    title_x=0.5,
)

fig.update_geos(
    projection_type="natural earth",
    showcountries=True,
    showcoastlines=False,
    showland=True,
    landcolor="#fdfdfd",
    oceancolor="#e5ecf6",
    showocean=True,
)

fig.show()
```

### Relationship Between Energy Supply and Protein Supply

<p style="text-align: justify;">
  This visualization tests whether higher <b>dietary energy availability</b> is associated
  with higher <b>protein supply</b> per person. We plot energy (kcal/capita/day) against
  protein (g/capita/day) to assess the relationship between food <b>quantity</b> and
  nutritional <b>quality</b>, where a positive trend suggests more calories generally come
  with more protein, while a wide spread indicates differences in diet composition.
</p>

```python
agg = (
    faostat_dataset[
        faostat_dataset["Indicator"].isin(["Energy supply", "Protein supply"])
    ]
    .groupby(["Area", "Indicator"], as_index=False)["Value"]
    .sum()
)

df = (
    agg.pivot(index="Area", columns="Indicator", values="Value")
       .rename(columns={
           "Energy supply": "Energy_kcal",
           "Protein supply": "Protein_g"
       })
       .reset_index()
       .dropna()
)

x = df["Energy_kcal"].to_numpy()
y = df["Protein_g"].to_numpy()

m, b = np.polyfit(x, y, 1)
r = np.corrcoef(x, y)[0, 1]
r2 = r ** 2

df["Fitted"] = m * df["Energy_kcal"] + b
df["Residual"] = df["Protein_g"] - df["Fitted"]

hi = df.nlargest(3, "Residual")
lo = df.nsmallest(3, "Residual")
ann_points = pd.concat([hi, lo])

x_line = np.linspace(df["Energy_kcal"].min(), df["Energy_kcal"].max(), 200)
y_line = m * x_line + b

fig = px.scatter(
    df,
    x="Energy_kcal",
    y="Protein_g",
    color="Residual",
    color_continuous_scale="RdBu_r",
    color_continuous_midpoint=0,
    hover_name="Area",
    hover_data={
        "Energy_kcal": ":.0f",
        "Protein_g": ":.0f",
        "Residual": ":+.1f"
    },
    labels={
        "Energy_kcal": "Energy (kcal/capita/day)",
        "Protein_g": "Protein (g/capita/day)",
        "Residual": "Protein vs trend (g)"
    },
    title=(
        "Energy vs Protein Supply by Country<br>"
        f"<sup>Linear fit: Protein = {m:.3f} × Energy + {b:.1f}  (R² = {r2:.2f})</sup>"
    ),
    template="plotly_white",
    opacity=0.85,
)

fig.add_trace(
    go.Scatter(
        x=x_line,
        y=y_line,
        mode="lines",
        name="Linear trend",
        line=dict(color="black", width=2, dash="dash"),
        hoverinfo="skip",
    )
)

for _, row in ann_points.iterrows():
    fig.add_annotation(
        x=row["Energy_kcal"],
        y=row["Protein_g"],
        text=row["Area"],
        showarrow=True,
        arrowhead=2,
        ax=20,
        ay=-20,
        bgcolor="rgba(255,255,255,0.85)",
        bordercolor="rgba(0,0,0,0.3)",
        borderwidth=1,
        font=dict(size=11),
    )

fig.update_traces(
    selector=dict(mode="markers"),
    marker=dict(
        size=9,
        line=dict(width=0.5, color="rgba(0,0,0,0.4)")
    ),
    hovertemplate=(
        "<b>%{hovertext}</b><br>"
        "Energy: %{x:,.0f} kcal/day<br>"
        "Protein: %{y:,.0f} g/day<br>"
        "Residual: %{marker.color:+.1f} g"
        "<extra></extra>"
    ),
)

fig.update_layout(
    height=700,
    title_x=0.5,
    margin=dict(l=60, r=80, t=90, b=80),

    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=-0.15,
        xanchor="center",
        x=0.5,
        bgcolor="rgba(255,255,255,0.8)",
        bordercolor="rgba(0,0,0,0.1)",
        borderwidth=1,
    ),

    coloraxis_colorbar=dict(
        title="Residual (g protein)",
        ticks="outside",
        x=1.02,
    ),
)

fig.show()
```

<p style="text-align: justify;">
  The scatter plot shows a <b>clear positive relationship</b> between energy availability
  and protein supply, but with notable spread. In general, higher-calorie countries provide
  more protein, yet outliers show that <b>calories alone do not guarantee protein adequacy</b>
  due to differences in diet composition.</p>

### Targeted Indicator–Food Group Selection

<p style="text-align: justify;">
  In this step, we create a <b>curated mapping</b> between each nutritional <b>Indicator</b>
  and the <b>most relevant Food Groups</b> that meaningfully contribute to that indicator.
  These food groups are <b>hand-picked</b> to focus the analysis on the strongest and most
  interpretable sources of each nutrient (e.g., protein from cereals/animal foods/legumes,
  Vitamin C from fruits and vegetables, fatty acids from fish). We then convert this mapping
  into a structured table and use it to <b>filter the FAOSTAT dataset</b> via an inner join.
</p>

```python
indicator_to_foodgroups = {
    'Protein supply': [
        'Cereals and their products',
        'Meat and meat products',
        'Milk and milk products',
        'Pulses, seeds and nuts and their products',
        'Fish, shellfish and their products',
    ],
    'Energy supply': [
        'Cereals and their products',
        'Fats and oils',
        'Sweets and sugars',
        'Meat and meat products',
        'Roots, tubers, plantains and their products'
    ],
    'Vitamin A supply (retinol activity equivalents)': [
        'Meat and meat products',
        'Vegetables and their products',
        'Milk and milk products',
        'Fats and oils',
        'Roots, tubers, plantains and their products'
    ],
    'Calcium supply': [
        'Milk and milk products',
        'Vegetables and their products',
        'Cereals and their products',
        'Pulses, seeds and nuts and their products'
    ],
    'Iron supply': [
        'Cereals and their products',
        'Meat and meat products',
        'Vegetables and their products',
        'Pulses, seeds and nuts and their products',
        'Roots, tubers, plantains and their products' 
    ],
    'Zinc supply': [
        'Cereals and their products',
        'Meat and meat products',
        'Milk and milk products',
        'Pulses, seeds and nuts and their products',
        'Roots, tubers, plantains and their products'
    ],
    'Vitamin C supply': [
        'Fruits and their products',
        'Vegetables and their products',
        'Roots, tubers, plantains and their products',
        'Beverages',
        'Milk and milk products'
    ],
    'Total saturated fatty acids supply': [
        'Fish, shellfish and their products',
    ],
    'Total monounsaturated fatty acids supply': [
        'Fish, shellfish and their products',
    ],
    'Selenium supply': [
        'Fish, shellfish and their products',
    ],
    'Vitamin B12 supply': [
        'Fish, shellfish and their products'
    ],
    'Vitamin B6 supply': [
        'Fish, shellfish and their products'
    ],
}
```

```python
indicator_food_mapping = (
    pd.Series(indicator_to_foodgroups)
      .explode()
      .reset_index()
      .rename(columns={'index': 'Indicator', 0: 'Food Group'})
)
```

```python
faostat_indicator_to_foodgroups_selected_dataset = (
    faostat_dataset_filtered
    .merge(indicator_food_mapping, on=['Indicator', 'Food Group'], how='inner')
    .reset_index(drop=True)
)
```

```python
faostat_indicator_to_foodgroups_selected_dataset
```

<p style="text-align: justify;">
  The final output dataset is therefore a <b>combined subset</b> that includes only the
  country-level observations where an <b>Indicator</b> matches one of its selected
  <b>Food Groups</b>. In other words, it represents a purposeful composition of
  <b>each indicator paired with the food groups defined above</b>, producing a cleaner and
  more decision-relevant dataset for downstream analysis.
</p>

### Final Indicator–Food Group Feature Table

<p style="text-align: justify;">
  In this step, we create a single descriptive feature label by combining the
  <code>Indicator</code> and <code>Food Group</code> fields into one column:
  <b>Indicator - Food Group</b>. This produces a clearer, self-explanatory structure where
  each row represents a specific nutrient indicator measured within a specific food group
  for a given country.
</p>

```python
faostat_indicator_to_foodgroups_selected_dataset['Indicator - Food Group'] = (
    faostat_indicator_to_foodgroups_selected_dataset['Indicator'] + ' - ' + faostat_indicator_to_foodgroups_selected_dataset['Food Group']
)
```

```python
final_faostat_indicator_to_foodgroups_selected_dataset = faostat_indicator_to_foodgroups_selected_dataset[['Area', 'Indicator - Food Group', 'Unit', 'Value']].reset_index(drop=True)
```

```python
final_faostat_indicator_to_foodgroups_selected_dataset
```

<p style="text-align: justify;">
  This final format is compact and model-ready, making it easier to filter, pivot, and compare
  countries across the selected indicator–food group combinations without carrying
  redundant rows
</p>

### Pivoting to a Model-Ready Country Matrix

<p style="text-align: justify;">
  In this step, we reshape the filtered indicator–food group dataset from a <b>long</b>
  format into a <b>wide</b> format using <code>pivot_table</code>. Each row represents one
  <b>country (Area)</b>, and each column represents a specific
  <b>Indicator - Food Group</b> feature, with the numeric <b>Value</b> placed in the
  corresponding cell. This structure is easier to use for comparison, correlation analysis,
  and machine-learning workflows.
</p>

```python
faostat_wide = (
    final_faostat_indicator_to_foodgroups_selected_dataset
    .pivot_table(
        index="Area",
        columns="Indicator - Food Group",
        values="Value",
        aggfunc="sum"
    )
    .reset_index()
)

faostat_wide.columns = [str(c) for c in faostat_wide.columns]
```

  country, producing a compact matrix (e.g., <b>189 rows × 40 columns</b>) where each column
  is directly interpretable.
</p>

### Lorenz Curve and Gini Index for TB Incidence Inequality

<p style="text-align: justify;">
  In this step, we quantify how unevenly <b>TB incidence</b> is distributed across countries
  using a <b>Lorenz curve</b>. We first sort country incidence values from lowest to highest,
  then plot the <b>cumulative share of countries</b> against the <b>cumulative share of total
  TB incidence</b>. The diagonal line represents <b>perfect equality</b>, while deviations
  below this line indicate increasing concentration of TB burden.
</p>

```python
tuberculosis_vals = worldbank_dataset_filtered_copy["tb_incidence"].dropna().values
tuberculosis_vals = tuberculosis_vals[tuberculosis_vals >= 0]  

def lorenz_curve(values: np.ndarray):
    """Return (p, L) where p is cumulative population share and L is cumulative incidence share."""
    x = np.sort(values)
    cumx = np.cumsum(x)
    total = cumx[-1]
    L = np.insert(cumx / total, 0, 0)    
    p = np.linspace(0, 1, len(L))         
    return p, L

def gini_from_lorenz(p: np.ndarray, L: np.ndarray) -> float:
    """Compute Gini from Lorenz curve via area under the curve."""
    area = np.trapz(L, p)
    gini = 1 - 2 * area
    return gini

p, L = lorenz_curve(tuberculosis_vals)
gini = gini_from_lorenz(p, L)

fig, ax = plt.subplots(figsize=(6, 6))

ax.plot([0, 1], [0, 1], linestyle="--", color="black", linewidth=1.5, label="Perfect equality")
ax.plot(p, L, color=COLORS["green"], linewidth=2.5, label="Lorenz curve")

ax.fill_between(p, L, 0, color=COLORS["green"], alpha=0.2)

ax.set_title("Lorenz curve for TB incidence", fontsize=13)
ax.set_xlabel("Cumulative share of countries")
ax.set_ylabel("Cumulative share of TB incidence")


ax.text(
    0.6, 0.15,
    f"Gini ≈ {gini:.2f}",
    fontsize=12,
    bbox=dict(boxstyle="round", facecolor="white", alpha=0.8),
    color=COLORS["green"]
)

ax.legend(loc="upper left")
ax.set_aspect("equal", "box")
plt.tight_layout()
plt.show()
```

<p style="text-align: justify;">
  The Lorenz curve lies substantially below the equality line, and the computed
  <b>Gini coefficient</b> (shown on the plot) is relatively high, indicating that TB
  incidence is <b>strongly concentrated</b> in a subset of countries. This reinforces the
  earlier rank-based results: global TB burden is not evenly spread, and we should
  account for this concentration when comparing regions and when interpreting global averages.
</p>

<p style="text-align: justify;">

  encodes indicator intensity by color, enabling quick identification of regional patterns and cross-country differences.

```python
def choropleth_with_hover(
    df,
    country_col,
    value_col,
    title,
    colorbar_title,
    scale="Viridis",
    fmt=".2f",
    unit_suffix=""
):
    data = df.copy()
    data[value_col] = pd.to_numeric(data[value_col], errors="coerce")

    fig = px.choropleth(
        data,
        locations=country_col,
        locationmode="country names",
        color=value_col,
        hover_name=country_col,
        custom_data=[value_col],
        color_continuous_scale=scale,
        title=title,
    )

    fig.update_traces(
        marker_line_color="black",
        marker_line_width=0.4,
        hovertemplate=(
            "<b>%{hovertext}</b><br>"
            f"{colorbar_title}: %{{customdata[0]:{fmt}}}{unit_suffix}"
            "<extra></extra>"
        ),
    )

    fig.update_layout(
        template="plotly_white",
        autosize=True,
        width=1100,
        height=600,
        margin=dict(l=0, r=20, t=60, b=0),
        title_x=0.5,
        font=dict(family="Arial, sans-serif", size=13, color="#333333"),
        hoverlabel=dict(
            bgcolor="white",
            bordercolor="#cccccc",
            font_size=12,
            font_family="Arial, sans-serif",
        ),
        coloraxis_colorbar=dict(
            title=colorbar_title,
            ticks="outside",
            ticklen=6,
            outlinewidth=0,
        ),
    )

    fig.update_geos(
        showcountries=True,
        countrycolor="black",
        showcoastlines=False,
        showland=True,
        landcolor="#fdfdfd",
        showocean=True,
        oceancolor="#e5ecf6",
        projection_type="natural earth",
    )

    return fig

fig_hdi = choropleth_with_hover(
    pivoted_hdi_clean,
    country_col="country",
    value_col="Human Development Index (value)",
    title="World Choropleth: Human Development Index (HDI) by Country",
    colorbar_title="HDI Value",
    scale="Viridis",
    fmt=".3f",
)
fig_hdi.show()

fig_life = choropleth_with_hover(
    pivoted_hdi_clean,
    country_col="country",
    value_col="Life Expectancy at Birth (years)",
    title="World Choropleth: Life Expectancy at Birth by Country",
    colorbar_title="Years",
    scale="Plasma",
    fmt=".1f",
    unit_suffix=" years",
)
fig_life.show()
```

### Data Integration and Preprocessing

<p style="text-align: justify;">
  The next step is to take the <b>FAOSTAT nutrition indicators</b>, <b>HDI development metrics</b>,
  <b>World Bank tuberculosis incidences</b>, and <b>BCG vaccination coverage</b> datasets and merge them into a single, analysis-ready dataset.
  Because each source uses different naming conventions and column schemas, the workflow prioritizes <b>schema alignment</b>
  (finding and renaming the correct key/value columns), <b>country-name standardization</b> (normalization + mapping dictionaries),
  and <b>merge diagnostics</b> The pipeline we use will also remove any duplicates, exclude
  non-country aggregates (e.g., regions/income groups), and standardizes the vaccine variable to a consistent percentage scale.
</p>

```python
faostat = faostat_wide
hdi     = pivoted_hdi_clean
tb      = worldbank_dataset_filtered
vaccine = pd.read_csv("./datasets/original_datasets/tuberculosis_vaccine.csv")
```

```python
def normalize_country(x):
    if pd.isna(x):
        return x
    s = str(x).strip()
    s = re.sub(r"\s+", " ", s)
    return s

def pick_first_existing_col(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

def show_unmatched(left_df, right_df, key="Area", title="Unmatched", top_n=30):
    L = set(left_df[key].dropna().unique())
    R = set(right_df[key].dropna().unique())
    only_L = sorted(L - R)
    only_R = sorted(R - L)
    print(f"\n--- {title} ---")
    print(f"Only in LEFT ({len(only_L)}): {only_L[:top_n]}")
    print(f"Only in RIGHT ({len(only_R)}): {only_R[:top_n]}")
```

```python
fa_col = pick_first_existing_col(faostat, ["Area", "Country", "country", "Entity", "DIM_GEO_NAME"])
if not fa_col:
    raise ValueError(f"FAOSTAT: не најдов колона за држава. Колони: {list(faostat.columns)}")
faostat = faostat.rename(columns={fa_col: "Area"})

hdi_col = pick_first_existing_col(hdi, ["country", "Area", "Country", "Entity"])
if not hdi_col:
    raise ValueError(f"HDI: не најдов колона за држава. Колони: {list(hdi.columns)}")
hdi = hdi.rename(columns={hdi_col: "Area"})

tb_geo_col = pick_first_existing_col(tb, ["DIM_GEO_NAME", "Area", "Country", "country"])
if not tb_geo_col:
    raise ValueError(f"TB: не најдов колона за држава. Колони: {list(tb.columns)}")
tb = tb.rename(columns={tb_geo_col: "Area"})

tb_val_col = pick_first_existing_col(tb, [
    "Tuberculosis incidence (per 100 000 population)",
    "Tuberculosis incidence (per 100,000 population)",
    "Tuberculosis incidence",
    "TB_incidence_2022",
])
if not tb_val_col:
    raise ValueError(f"TB: не најдов incidence колона. Колони: {list(tb.columns)}")
tb = tb.rename(columns={tb_val_col: "TB_incidence_2022"})

vac_geo_col = pick_first_existing_col(vaccine, ["Entity", "Area", "Country", "country"])
if not vac_geo_col:
    raise ValueError(f"Vaccine: не најдов колона за држава. Колони: {list(vaccine.columns)}")
vaccine = vaccine.rename(columns={vac_geo_col: "Area"})

year_col = pick_first_existing_col(vaccine, ["Year", "year"])
if not year_col:
    raise ValueError(f"Vaccine: не најдов колона Year. Колони: {list(vaccine.columns)}")
vaccine = vaccine.rename(columns={year_col: "Year"})

vac_val_col = pick_first_existing_col(vaccine, [
    "Share of newborns vaccinated against tuberculosis",
    "Share of newborns vaccinated against tuberculosis (%)",
    "BCG_coverage_pct",
])
if not vac_val_col:
    raise ValueError(f"Vaccine: не најдов BCG share колона. Колони: {list(vaccine.columns)}")
```

```python
for df in [faostat, hdi, tb, vaccine]:
    df["Area"] = df["Area"].apply(normalize_country)
```

```python
exclude_tb_non_countries = {
    "African Region", "Eastern Mediterranean Region", "European Region",
    "Region of the Americas", "South-East Asia Region", "Western Pacific Region",
    "Global", "GLOBAL"
}
tb = tb[~tb["Area"].isin(exclude_tb_non_countries)].copy()
tb = tb[["Area", "TB_incidence_2022"]].copy()
```

```python
vaccine_2022 = vaccine[vaccine["Year"] == 2022].copy()
vaccine_2022 = vaccine_2022.rename(columns={vac_val_col: "BCG_coverage_pct"})
vaccine_2022 = vaccine_2022[["Area", "BCG_coverage_pct"]].copy()

vaccine_2022 = vaccine_2022.groupby("Area", as_index=False)["BCG_coverage_pct"].mean()

mx = vaccine_2022["BCG_coverage_pct"].max(skipna=True)
if pd.notna(mx) and mx <= 1.0:
    vaccine_2022["BCG_coverage_pct"] = vaccine_2022["BCG_coverage_pct"] * 100.0
```

```python
map_tb = {
    "United Kingdom": "United Kingdom of Great Britain and Northern Ireland",
}

map_hdi = {
    "United States": "United States of America",
    "United Kingdom": "United Kingdom of Great Britain and Northern Ireland",
    "Turkey": "Türkiye",
    "Tanzania (United Republic of)": "United Republic of Tanzania",
    "Moldova (Republic of)": "Republic of Moldova",
    "Korea (Republic of)": "Republic of Korea",
    "Netherlands": "Netherlands (Kingdom of the)",
    "Congo (Democratic Republic of the)": "Democratic Republic of the Congo",
    "Eswatini (Kingdom of)": "Eswatini",
}

map_vaccine = {
    "United Kingdom": "United Kingdom of Great Britain and Northern Ireland",
    "United States": "United States of America",
    "Turkey": "Türkiye",
    "Tanzania": "United Republic of Tanzania",
    "South Korea": "Republic of Korea",
    "Russia": "Russian Federation",
    "Vietnam": "Viet Nam",
    "Cape Verde": "Cabo Verde",
    "East Timor": "Timor-Leste",
    "Syria": "Syrian Arab Republic",
    "Laos": "Lao People's Democratic Republic",
    "Democratic Republic of Congo": "Democratic Republic of the Congo",
}

tb["Area"] = tb["Area"].replace(map_tb)
hdi["Area"] = hdi["Area"].replace(map_hdi)
vaccine_2022["Area"] = vaccine_2022["Area"].replace(map_vaccine)

for df in [tb, hdi, vaccine_2022]:
    df["Area"] = df["Area"].apply(normalize_country)

show_unmatched(faostat, tb, key="Area", title="Before merge: FAOSTAT vs TB")
show_unmatched(faostat, hdi, key="Area", title="Before merge: FAOSTAT vs HDI")
show_unmatched(faostat, vaccine_2022, key="Area", title="Before merge: FAOSTAT vs Vaccine(2022)")
```

```python
faostat = faostat.drop_duplicates(subset=["Area"]).copy()
tb      = tb.drop_duplicates(subset=["Area"]).copy()
hdi     = hdi.drop_duplicates(subset=["Area"]).copy()
vaccine_2022 = vaccine_2022.drop_duplicates(subset=["Area"]).copy()
```

```python
faostat_tb = faostat.merge(tb, on="Area", how="inner")
faostat_tb_hdi = faostat_tb.merge(hdi, on="Area", how="inner")
final_df = faostat_tb_hdi.merge(vaccine_2022, on="Area", how="left")
```

```python
final_df["BCG_was_missing"] = final_df["BCG_coverage_pct"].isna()

TARGET_COL = "TB_incidence_2022"
if TARGET_COL not in final_df.columns:
    raise ValueError(f"Target колоната '{TARGET_COL}' не постои по merge. Провери TB rename.")

numeric_cols = final_df.select_dtypes(include=[np.number]).columns.tolist()

impute_cols = [c for c in numeric_cols if c != TARGET_COL]

imputer = IterativeImputer(
    random_state=42,
    max_iter=20,
    initial_strategy="median"
)

X_imp = imputer.fit_transform(final_df[impute_cols])
final_df[impute_cols] = X_imp

final_df["BCG_coverage_pct"] = final_df["BCG_coverage_pct"].clip(0, 100)
```

```python
cols = list(final_df.columns)
middle_cols = [c for c in cols if c not in ["Area", TARGET_COL]]
final_df = final_df[["Area"] + middle_cols + [TARGET_COL]]
```

```python
print("Final shape:", final_df.shape)

dup_areas = int(final_df["Area"].duplicated().sum())
print("Duplicate Area rows:", dup_areas)

missing_by_col = final_df.isna().sum().sort_values(ascending=False)
missing_nonzero = missing_by_col[missing_by_col > 0]
print("Columns with missing values:", int((missing_by_col > 0).sum()))
if not missing_nonzero.empty:
    display(missing_nonzero.to_frame("missing_count"))

print("\nColumns (last 5) to confirm target is last:")
print(final_df.columns[-5:])
```

### Model Training Setup and Ridge Regression

<p style="text-align: justify;">
  In this step, we prepare the <b>final integrated dataset</b> for predictive modeling by separating the target variable
  (<b>TB_incidence_2022</b>) from the explanatory features, while keeping country names only for reference. Because TB incidence
  is highly skewed, the target is also transformed using <code>log1p</code> to stabilize variance and reduce the influence of extreme
  values. The dataset is then split into <b>training</b> and <b>test</b> sets using a fixed random seed to ensure reproducibility.
</p>

```python
RANDOM_SEED = 42
TEST_SIZE = 0.20
results = []
COUNTRY_COL = "Area"
TARGET_COL = "TB_incidence_2022"
```

```python
required_cols = [COUNTRY_COL, TARGET_COL]
countries_all = final_df[COUNTRY_COL].astype(str).copy()
y_target_raw = final_df[TARGET_COL].astype(float).to_numpy()
y_target_log = np.log1p(y_target_raw)
X_features_df = final_df.drop(columns=required_cols).copy()
```

```python
print("X_features_df shape:", X_features_df.shape)
print("y_target_raw shape:", y_target_raw.shape)
print("y_target_log shape:", y_target_log.shape)
print("y_target_raw min/max:", float(np.min(y_target_raw)), "/", float(np.max(y_target_raw)))
```

```python
X_train, X_test, y_train_raw, y_test_raw, countries_train, countries_test = train_test_split(
    X_features_df,
    y_target_raw,
    countries_all,
    test_size=TEST_SIZE,
    random_state=RANDOM_SEED,
    shuffle=True
)
```

```python
y_train_log = np.log1p(y_train_raw)
y_test_log = np.log1p(y_test_raw)

print("X_train_df:", X_train.shape, "| X_test_df:", X_test.shape)
print("y_train_raw:", y_train_raw.shape, "| y_test_raw:", y_test_raw.shape)
```

```python
ridge_pipeline_log = Pipeline(steps=[
    ("scaler", StandardScaler()),
    ("regressor", Ridge())
])
```

```python
alpha_values = np.logspace(-3, 3, 13)
param_grid = {"regressor__alpha": alpha_values}

cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)

ridge_cv_log = GridSearchCV(
    estimator=ridge_pipeline_log,
    param_grid=param_grid,
    cv=cv,
    scoring="neg_mean_absolute_error",
    n_jobs=-1
)
```

```python
ridge_cv_log.fit(X_train, y_train_log)

print("Best alpha (Ridge LOG):", ridge_cv_log.best_params_["regressor__alpha"])
print("Best CV MAE (log-scale):", -ridge_cv_log.best_score_)
```

```python
best_ridge_log = ridge_cv_log.best_estimator_
y_pred_log_ridge = best_ridge_log.predict(X_test)

y_pred_log_ridge = np.clip(y_pred_log_ridge, a_min=0.0, a_max=None)
```

```python
res_ridge_log = {
    "Model": f"Ridge LOG alpha={ridge_cv_log.best_params_['regressor__alpha']}",
    "MAE_log": mean_absolute_error(y_test_log, y_pred_log_ridge),
    "RMSE_log": np.sqrt(mean_squared_error(y_test_log, y_pred_log_ridge)),
    "R2_log": r2_score(y_test_log, y_pred_log_ridge),
}
results.append(res_ridge_log)
```

```python
results_df = (
    pd.DataFrame(results)[["Model", "MAE_log", "RMSE_log", "R2_log"]]
    .sort_values(by="RMSE_log", ascending=True)
    .reset_index(drop=True)
)

results_df
```

<p style="text-align: justify;">
  A <b>Ridge Regression</b> model is trained using a pipeline that standardizes features (<code>StandardScaler</code>) and fits a ridge
  regressor. To select the best regularization strength, a grid of <code>alpha</code> values is evaluated with <b>5-fold cross-validation</b>,
  optimizing for <b>mean absolute error (MAE)</b> in the log scale. The best model is then tested on the held-out set, and performance is
  summarized using <b>MAE</b>, <b>RMSE</b>, and <b>R²</b> (all in log scale) inside <code>results_df</code>.
</p>

<p style="text-align: justify;">
  Overall, this workflow produces a reproducible baseline model that balances fit and generalization through regularization.
  The log transformation ensures that prediction errors are evaluated more fairly across <b>low- and high-incidence countries</b>, and the
  final results table provides a clear benchmark for comparing Ridge against other regression models in the future.
</p>

### Ridge SHAP Summary Plot

<p style="text-align: justify;">
  In this step, we generate a <b>SHAP summary (beeswarm) plot</b> for the trained Ridge model to interpret how each feature contributes
  to predictions of <b>TB incidence</b>. The code is focused on <b>visual clarity</b>: it applies smaller fonts, a shorter figure height,
  and reduced point sizes so the complete set of features can be displayed in a single, scroll-free figure.
</p>

```python
scaler = best_ridge_log.named_steps["scaler"]
ridge  = best_ridge_log.named_steps["regressor"]

feature_names = list(X_train.columns)

X_train_scaled = scaler.transform(X_train)
X_test_scaled  = scaler.transform(X_test)

X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=feature_names, index=X_train.index)
X_test_scaled_df  = pd.DataFrame(X_test_scaled,  columns=feature_names, index=X_test.index)

def short_label(name: str, keep_prefix_chars=22, keep_suffix_chars=18):
    s = str(name)
    if len(s) <= keep_prefix_chars + keep_suffix_chars + 5:
        return s
    return s[:keep_prefix_chars].rstrip() + " ... " + s[-keep_suffix_chars:].lstrip()

short_names = [short_label(c) for c in feature_names]
name_map_df = pd.DataFrame({"short": short_names, "full": feature_names})

X_train_scaled_df_plot = X_train_scaled_df.copy()
X_test_scaled_df_plot  = X_test_scaled_df.copy()
X_train_scaled_df_plot.columns = short_names
X_test_scaled_df_plot.columns  = short_names

coef = np.asarray(ridge.coef_).ravel()
baseline = X_train_scaled_df.values.mean(axis=0)

shap_values_np = (X_test_scaled_df.values - baseline) * coef

shap_values = shap.Explanation(
    values=shap_values_np,
    base_values=np.repeat(float(ridge.intercept_), X_test_scaled_df.shape[0]),
    data=X_test_scaled_df.values,
    feature_names=short_names
)

n_features = X_test_scaled_df.shape[1]
max_display = n_features  # ALL features

plt.close("all")

plt.rcParams.update({
    "font.size": 7,
    "axes.titlesize": 9,
    "axes.labelsize": 8,
    "xtick.labelsize": 7,
    "ytick.labelsize": 6,
    "figure.dpi": 220,
    "savefig.dpi": 300,
})

fig_w = 12.5

row_h = 0.16
cap_h = None
fig_h = row_h * n_features + 1.5
fig_h = max(5.0, fig_h)
if cap_h is not None:
    fig_h = min(cap_h, fig_h)

fig = plt.figure(figsize=(fig_w, fig_h))

try:
    shap.plots.beeswarm(
        shap_values,
        max_display=max_display,
        show=False,
        plot_size=(fig_w, fig_h),
        s=10,
        alpha=0.7
    )
except Exception:
    shap.summary_plot(
        shap_values,
        X_test_scaled_df_plot,
        show=False,
        max_display=max_display,
        plot_size=(fig_w, fig_h),
        dot_size=10,
        alpha=0.7
    )

ax = plt.gca()

ax.set_title(f"Ridge SHAP summary (n={X_test_scaled_df_plot.shape[0]:,}, p={n_features})", pad=6)
ax.set_xlabel("SHAP value (impact on model output)", labelpad=6)

ax.axvline(0, linewidth=0.7, alpha=0.55)
ax.grid(True, axis="x", linestyle="--", linewidth=0.5, alpha=0.30)
ax.set_axisbelow(True)

ax.xaxis.set_major_locator(mticker.MaxNLocator(nbins=7))
ax.xaxis.set_major_formatter(mticker.FormatStrFormatter("%.3g"))

ax.tick_params(axis="y", labelsize=6, pad=2)

max_len = max(len(s) for s in short_names) if short_names else 0
left = 0.22 if max_len < 28 else 0.30 if max_len < 40 else 0.36
fig.subplots_adjust(left=left, right=0.96, top=0.92, bottom=0.10)

out_png = "shap_ridge_summary_ALL_enhanced.png"
plt.show()
```

<p style="text-align: justify;">
  The plot ranks variables by their overall importance (mean absolute SHAP value). Each dot represents one country in the test set:
  the <b>x-axis</b> shows the direction and magnitude of impact on the model output, while color encodes whether the feature value is
  relatively <b>high</b> or <b>low</b>. A vertical line at <b>0</b> separates features pushing predictions higher vs. lower.
</p>

###  Identifying the Strongest Predictor with SHAP

<p style="text-align: justify;">
  In this step, SHAP values are used to determine the <b>single most influential feature</b> in the Ridge model. We compute the
  <b>mean absolute SHAP value</b> for every predictor across the test set, which measures how strongly each variable contributes to the
  model’s predictions on average (regardless of direction). The feature with the highest mean(|SHAP|) is reported as the
  <b>best indicator</b> because it has the largest overall impact on predicted TB incidence.
</p>

```python
mean_abs = np.abs(shap_values.values).mean(axis=0)
top_idx = int(mean_abs.argmax())
top_feature = feature_names[top_idx]

print("Top SHAP feature (best indicator):", top_feature)
print("Mean(|SHAP|):", float(mean_abs[top_idx]))

shap.dependence_plot(top_feature, shap_values.values, X_test_scaled_df, feature_names=feature_names)
```

<p style="text-align: justify;">
  After selecting this top feature, a <b>SHAP dependence plot</b> is generated. This plot shows how changes in the chosen indicator’s
  (scaled) values relate to its SHAP contribution, making it easier to interpret whether higher values of the feature tend to
  <b>increase</b> or <b>decrease</b> the model’s output, and whether the relationship appears linear, non-linear, or influenced by outliers.
</p>
